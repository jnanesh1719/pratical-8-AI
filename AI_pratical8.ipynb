```python
# pract_8
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB

messages = [
    "You won 1 lakh! Claim now",
    "Click here to win a free iPhone",
    "Urgent! Get a free cash prize",
    "Congratulations! You are a lucky winner",
    "Hi, how are you?",
    "Please attend class tomorrow",
    "Your assignment is due today",
    "Good morning, have a nice day"
]

labels = [
    "spam",
    "spam",
    "spam",
    "spam",
    "not spam",
    "not spam",
    "not spam",
    "not spam"
]

# Creating the CountVectorizer
vectorizer = CountVectorizer()

# Transforming messages into numerical data
X = vectorizer.fit_transform(messages)

print("Vocabulary:", vectorizer.vocabulary_)
print(X.toarray())

# Creating and training the model
model = MultinomialNB()
model.fit(X, labels)

# Predicting a new message
new_message = ["Win cash today"]
new_X = vectorizer.transform(new_message)
prediction = model.predict(new_X)

print("Message:", new_message[0])
print("Prediction:", prediction[0])

# Taking user input
user_message = input("Enter a message: ")
user_X = vectorizer.transform([user_message])
prediction = model.predict(user_X)

print("Prediction:", prediction[0])
```
